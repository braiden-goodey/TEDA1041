```python
import pandas as pd
import requests
from bs4 import BeautifulSoup
```

```python
response = requests.get('https://btech-data-analytics.github.io/bridgerland-technical-college/bookstore.html').text
```

```python
soup = BeautifulSoup(response, 'html.parser')
```

```python
ISBN =[]
Title = []
Author = []
Language = []
Pages = []
Topic = []
Price = []

for employee in soup.find ('table').find_all('tr', class_='book'):
  ISBN.append (employee.find_all('td')[0].text)
  Title.append (employee.find_all('td')[1].text)
  Author.append (employee.find_all('td')[2].text)
  Language.append (employee.find_all('td')[3].text)
  Pages.append (employee.find_all('td')[4].text)
  Topic.append (employee.find_all('td')[5].text)
  Price.append (employee.find_all('td')[6].text)
```

```python
df = pd.DataFrame({
    'ISBN': ISBN,
    'Title': Title,
    'Author': Author,
    'Language': Language,
    'Pages': Pages,
    'Topic': Topic,
    'Price': Price
})
df.head()
```

```python
#Q1
df['Author'].value_counts()
```

```python
#Q2
df['Topic'].value_counts()
```

```python
#Q3
df['Price'] = df['Price'].str.replace('$', '', regex=False).astype(float)
df.groupby('Topic')['Price'].mean()
```

```python
#Q4
df['Pages'] = df['Pages'].astype(int)
df.groupby('Topic')['Pages'].mean()
```
