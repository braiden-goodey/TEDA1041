```python
#Q14
df["Tax_Amount"] = (df["UnitPrice"] * df["Quantity"] * df["TaxRate"]/100)
df.head()
```

```python
#Q15
df["Extended_Price"] = (df["UnitPrice"] * df["Quantity"]) + df["Tax_Amount"]
df.head()
```

```python
df.describe()
```

```python
#Q18-19
#Use Comments to run either Head, PZS or QZS
average_price = df['Extended_Price'].mean()
std_dev_price = df['Extended_Price'].std()
average_Quantity = df['Quantity'].mean()
std_dev_Quantity = df['Quantity'].std()
df['Price_Z-Score'] = ((df['Extended_Price']- average_price)/ std_dev_price)
df['Quantity_Z-Score'] = ((df['Quantity']- average_Quantity)/ std_dev_Quantity)
PZS_POS=(df['Price_Z-Score'] >= 3)
PZS_NEG=(df['Price_Z-Score'] <= -3)
QZS_POS=(df['Quantity_Z-Score'] >= 3)
QZS_NEG=(df['Quantity_Z-Score'] <= -3)
df.head()
(PZS_POS + PZS_NEG).sum()
# (QZS_POS + QZS_NEG).sum()
```
