```python
import seaborn as sns
import pandas as pd
import matplotlib.pyplot as plt
df = pd.read_csv('housing.csv')
df.head()
```

```python
sns.scatterplot(data=df,x='Landsize', y='Price', hue='Type' )
plt.show()
```

```python
avg_price = df.groupby('YearBuilt')['Price'].mean().reset_index()
avg_price.rename(columns={'Price': 'avg_price'}, inplace=True)
sns.lineplot(data=avg_price, x='YearBuilt', y='avg_price')
plt.show()
```

```python
avg_price = df.groupby('Regionname')['Price'].mean().reset_index()
avg_price.rename(columns={'Price': 'avg_price'}, inplace=True)
sns.barplot(data=avg_price, x='Regionname', y='avg_price')
plt.xticks(rotation=45)
plt.show()
```

```python
sns.histplot(data=df, x='Price')
plt.show()
```

```python
sns.boxplot(data=df, x='YearBuilt', y= 'Regionname')
plt.show()
```
