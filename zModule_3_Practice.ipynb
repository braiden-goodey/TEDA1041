```python
import pandas as pd
import requests
from bs4 import BeautifulSoup
```

```python
response = requests.get('https://btech-data-analytics.github.io/bridgerland-technical-college/employees_3.html').text
```

```python
soup = BeautifulSoup(response, 'html.parser')
```

```python
# Access Table, Access Class, Access Employee, Get Text only
soup.find ('table').find_all('tr', class_='employee')[0].find_all('td')[1].text
```

```python
#Sorts lists of variables
id = []
name = []
age = []

for employee in soup.find ('table').find_all('tr', class_='employee'):
  id.append (employee.find_all('td')[0].text)
  name.append (employee.find_all('td')[1].text)
  age.append (employee.find_all('td')[2].text)
```

```python
#Prints vairbale
name
```

```python
#Created data frame from web data....SCRAPING
df = pd.DataFrame({
    'id': id,
    'name': name,
    'age': age
})
df.head()
```

```python
df['age_num'] = [float(s) for s in df['age']]
Ave_Age = df['age_num'].mean()
STD_Age = round(df['age_num'].std(),2)
print(Ave_Age)
print(STD_Age)
```
